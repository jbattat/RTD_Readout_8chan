```python
#define anything you need to, ie commands

#this is where I define my command and transcribe it into whatever language the device understands
cmd = 't'.encode('utf_8')

#this will parse the response into something readable
def parse_response(response):
    try:
        #decodes response and strips \r\n
        response = response.decode('utf_8', errors='replace').strip('\r').strip('\n').strip() 
        #print(f"Raw Data String: {response}")

        #split the string by delimiter
        values = response.split(',')
        
        if len(values) < 4: #put the number of expected measurements
            raise ValueError("Not enough data fields")

        #seperates the data as floats
        A = float(values[0])
        B = float(values[1])
        C = float(values[2])
        D = float(values[3])
        E = float(values[4])
        F = float(values[5])
        G = float(values[6])
        H = float(values[7])
        return A, B, C, D, E, F, G, H

    except (ValueError, IndexError) as e:
        print(f"Parsing error: {e}")
        #return np.nan, np.nan, np.nan
```

```python
#use this cell to test if you can send a command and recieve a response once

sock.sendall(cmd) #CHANGED FROM SERIAL #send encoded command
#print(cmd)

time.sleep(1)

try:
    #print('in_waiting')
    response = sock.recv(1024) #CHANGED FROM SERIAL #have to set argument for number of bytes
    print('a')
    print(response)
    if response:
        print('a')
        mytuple = parse_response(response) #puts returned measurements into a tuple
        (A, B, C, D, E, F, G, H) = mytuple
        now_time = dt.datetime.now() #pulls current time
        current_time = str(now_time.strftime('%Y%m%dT%H%M%S.%f')[:-3])
        print(f"Measurements at {current_time}: \n T1: {A} \n T2: {B} \n T3: {C} \n T4: {D} \n T5: {E} \n T6: {F} \n T7: {G} \n T8: {H} ")
    else:
        print('No response')
except:
    print('No inwaiting')
```

```python
#use this cell to test if you can repeat sending and receiving

while True:
    sock.sendall(cmd) #send encoded command
    
    time.sleep(.1)

    response = sock.recv(1024) #have to set argument for number of bytes
    if response:
        #print(response)
        mytuple = parse_response(response) #puts returned measurements into a tuple
        (A, B, C, D, E, F, G, H) = mytuple
        now_time = dt.datetime.now() #pulls current time
        current_time = str(now_time.strftime('%Y%m%dT%H%M%S.%f')[:-3])
        print(f"Measurements at {current_time}: \n T1: {A} \n T2: {B} \n T3: {C} \n T4: {D} \n T5: {E} \n T6: {F} \n T7: {G} \n T8: {H} ")
    else:
        print('No response')

    time.sleep(.1)
```
